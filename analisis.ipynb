# Análisis de sensores industriales

Examen práctico — Manejo Masivo de Datos. Los datos del archivo `sensores_industriales.csv` son **simulados**. Umbral de alerta (regla didáctica del examen): temperatura mayor que 85 °C.

```python
from pathlib import Path
import pandas as pd

umbral = 85
df = pd.read_csv(Path('data') / 'sensores_industriales.csv')
df.head()
```

## 1. Cantidad de registros y de sensores distintos

```python
print('Registros:', len(df))
print('Sensores distintos:', df['id_sensor'].nunique())
```

## 2. Temperatura promedio por planta

```python
df.groupby('planta')['temperatura_c'].mean().round(2)
```

## 3. Temperatura máxima, sensor y fecha

Se muestran todos los registros empatados con el valor máximo.

```python
tmax = df['temperatura_c'].max()
print('Temperatura máxima:', tmax)
df[df['temperatura_c'] == tmax][['id_sensor', 'fecha_hora', 'temperatura_c']].head(10)
```

## 4. Lecturas con temperatura mayor que 85 °C

```python
alertas = df[df['temperatura_c'] > umbral]
print('Alertas de temperatura:', len(alertas))
```

## 5. Planta con más alertas

Si hay empate se muestran todas las plantas empatadas.

```python
conteo = alertas['planta'].value_counts()
print(conteo)
conteo[conteo == conteo.max()]
```

## 6. Exportar alertas a `resultados/alertas.csv`

```python
Path('resultados').mkdir(exist_ok=True)
alertas.to_csv(Path('resultados') / 'alertas.csv', index=False)
pd.read_csv(Path('resultados') / 'alertas.csv').head()
```
